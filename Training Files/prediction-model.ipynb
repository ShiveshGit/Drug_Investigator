```python
import torch
```

```python
model = torch.load("./results/trained_model.pkl")
model
```

```python
from pykeen.triples import TriplesFactory
triples_factory = TriplesFactory.from_path_binary("./results/training_triples/")
print(triples_factory.num_entities)  # Number of entities
print(triples_factory.num_relations)  # Number of relations
```

```python
# triples_factory
```

```python
from pykeen.predict import predict_target
pred = predict_target(
    model = model,
    head="http://purl.obolibrary.org/obo/DOID_13099",
    relation="Parents",
    triples_factory = triples_factory.training
)
```

```python
df = pred.df
df
```

```python
df.nlargest(n=10, columns="score")
```
