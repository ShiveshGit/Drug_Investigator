```python
df2.dropna(axis=0,how='any',inplace=True)   
df2.to_csv("DTO_cleaned4.csv", index=False)
```

```python
data4=pd.read_csv("DTO_cleaned4.csv") 
```

```python
for idx, i in enumerate(df2['object']):
    if "|" in str(i):  # Check if the object contains '|'
        arr = i.split("|")  # Split the object string by '|'
        for k, val in enumerate(arr):  # Iterate over the split values
            if k == 0:
                # Append the first split value with original subject and predicate
                data4['subject'].append(df2['subject'][idx])
                data4['predicate'].append(df2['predicate'][idx])
                data4['object'].append(val)
            else:
                # Append subsequent split values with modified predicate
                data4['subject'].append(df2['subject'][idx])
                data4['predicate'].append(f"{df2['predicate'][idx]}_{k-1}")
                data4['object'].append(val)
    else:
        # If no '|', append the original row values
        data4['subject'].append(df2['subject'][idx])
        data4['predicate'].append(df2['predicate'][idx])
        data4['object'].append(i)

# Creating a new DataFrame from the transformed dictionary
df3 = pd.DataFrame(data4)

# Saving the transformed DataFrame to a CSV file
df3.to_csv("DTO_cleaned5.csv", index=False)

# Displaying the resulting DataFrame (optional)
print(df3)
```
